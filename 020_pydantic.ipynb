```python
from pydantic import BaseModel, ValidationError

class Person(BaseModel):
    first_name: str
    last_name: str
    age: int = 0

p = Person(first_name='John', last_name='Smith', age=42)
print(f"{p=}")
print(f"{p.model_fields=}")
```

```python
# export as json
p_json_sting = p.model_dump_json()
print(p_json_sting)

# import from json
p2 = Person.model_validate_json(p_json_sting)
print(f"{p2=}")
print(type(p2))
```

```python
class Worker(Person):
    salary: float

w = Worker(first_name='John', last_name='Smith', salary=1000.0)
print(f"{w=}")

# export as json
w_json_sting = w.model_dump_json()
print(w_json_sting)

# import from json
w2 = Worker.model_validate_json(w_json_sting)
print(f"{w2=}")
print(type(w2))
```

```python
class Student(Person):
    grade: int | None = None

s = Student(first_name='John', last_name='Smith', age=42, grade=5)
print(f"{s=}")
print(f"{s.model_fields=}")

# export as json
s_json_sting = s.model_dump_json()
print(s_json_sting)

# import from json
s2 = Student.model_validate_json(s_json_sting)
print(f"{s2=}")
print(type(s2))
```

```python
p3 = Person.model_validate_json(s_json_sting)
print(f"{p3=}")
print(type(p3))

try:
    w3 = Worker.model_validate_json(s_json_sting)
except ValidationError as e:
    print(f"{e=}")
    print(f"{e.errors()=}")
    print(f"{e.json()=}")
```

```python
class Mathematician(Person):
    favorite_numbers: list[int] = []  # Note that this default value is allowed in pydantic

m1 = Mathematician(first_name='John', last_name='Smith')
m1.favorite_numbers.append(42)

m2 = Mathematician(first_name='Karl', last_name='Gauss')
m2.favorite_numbers.append(35)

print(f"{m1=}")
print(f"{m1.model_fields=}")
print(f"{m2=}")
print(f"{m2.model_fields=}")
```

```python
# import dataclasses

# @dataclasses.dataclass
# class Mathematician2:
#     favorite_numbers: list[int] = []  # This is not allowed!!!
```

```python
import dataclasses

@dataclasses.dataclass
class Mathematician2:
    favorite_numbers: list[int] = dataclasses.field(default_factory=list)


m3 = Mathematician2()
m3.favorite_numbers.append(42)
m4 = Mathematician2()
m4.favorite_numbers.append(35)

print(f"{m3=}")
print(f"{m4=}")
```
